Bài 2B: Viết hàm zigzag_numbers(n) trả về một chuỗi biểu diễn ma trận n x n chứa các số từ 1 đến n^2. Các hàng cách nhau bởi ký tự xuống dòng. Với chỉ số hàng bắt đầu từ 0: hàng chẵn đi từ trái sang phải, hàng lẻ đi từ phải sang trái. Các số trong cùng một hàng cách nhau bởi một dấu cách. 
n = 2
1 2
4 3
n= 4
1 2 3 4
8 7 6 5
9 10 11 12
16 15 14 13

```python
def zigzag_numbers(matrix):
    kq = ""
    for i in range(matrix):
        if i%2==0:
            hang = list(range(i*n+1,(i+1)*n+1))
        else:
            hang = list(range((i+1)*n,i*n+1-1,-1))
        for j in range(len(hang)):
            kq += str(hang[j])
            if j<len(hang)-1:
                kq += " "
        if i< matrix-1:
            kq += "\n"
    return kq
n = int(input())
print(zigzag_numbers(n))


```

Bài 3: Trả về cột (chỉ số bắt đầu từ 0) có tổng cột lớn nhất trong ma trận n x m. 
Viết hàm column_with_max_sum(mat) trả về chỉ số bắt đầu từ 0 của cột có tổng lớn nhất. 
Nếu có nhiều cột cùng tổng lớn nhất, trả về cột có chỉ số nhỏ nhất. - Dùng vòng lặp để tính tổng từng cột. - Không dùng tuple hoặc dictionary. 
n = 2
m = 3
mat = [[1, 3, 5], 
       [6, 2, -9]]


```python
def column_with_max_sum(matrix):
    sum_max = -1000
    col_max = -1
    n = len(matrix)
    m = len(matrix[0])
    for j in range(m):
        s= 0 
        for i in range(n):
            s+= matrix[i][j]
        if s>sum_max:
            sum_max = s
            col_max = j
    return col_max, sum_max
n,m = map(int, input().split())
matrix = []
for i in range(n):
    row = list(map(int, input().split()))
    matrix.append(row)
for i in range(n):
    print(*matrix[i])
print(column_with_max_sum(matrix))

        
```

Bài 4: Trả về hàng (chỉ số bắt đầu từ 0) có tổng hàng nhỏ nhất trong ma trận n x m. 
Viết hàm row_with_min_sum(mat) trả về chỉ số bắt đầu từ 0 của hàng có tổng nhỏ nhất. 
Nếu có nhiều hàng cùng tổng nhỏ nhất, trả về hàng có chỉ số lớn nhất. - Dùng vòng lặp để tính tổng từng hàng. -. 
n = 2
m = 3
mat = [[1, 3, 5], 
       [6, 2, -9]]


```python
def row_with_min_sum(mat):
    index_min_row = -1
    min_sum_row = float('inf')
    n = len(mat)
    m = len(mat[0])
    for i in range(n):
        s = 0
        for j in range(m):
            s += mat[i][j]
        if s<=min_sum_row:
            min_sum_row = s
            index_min_row = i
    return index_min_row
n = int(input())
m = int(input())
matrix = []
for i in range(n):
    row = list(map(int, input().split()))
    matrix.append(row)
print(row_with_min_sum(matrix))

        

```

Bài 6: Cho ma trận nxm số nguyên, thêm vào một cột tại vị trí k.
