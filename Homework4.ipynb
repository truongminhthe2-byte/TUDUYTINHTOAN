## W6A19 
Viết hàm đưa vào 1 list số nguyên, tìm và trả về vị trí có giá trị lớn nhất 
trong list 

```python
def vi_tri_max(lst):
    max_list = lst[0]
    max_vi_tri = 0
    for i in range(len(lst)):
        if lst[i]>max_list:
            max_list = lst[i]
            max_vi_tri = i
    return max_vi_tri
lst = list(map(int, input().split()))
print(vi_tri_max(lst))
```

## W6A21 
Nhập một dòng gồm các chuỗi (tách bởi dấu cách). Tạo dictionary ánh xạ 
chuỗi -> tuple các vị trí (0-based) mà chuỗi đó xuất hiện 

```python
chuoi = input().split()
dct = {}
for i, v in enumerate(chuoi):
    if v not in dct:
        dct[v] = [i]
    else:
        dct[v].append(i)
for k in dct:
    dct[k] = tuple(dct[k])
print(dct)

```

## W6A23 VIết hàm xóa đi các số <x có trong mảng 

```python
def remove_below_x(lst,x):
    new_lst = []
    for i in lst:
        if i>=x:
            new_lst.append(i)
    return new_lst
lst = list(map(int, input().split()))
x = int(input())
print(remove_below_x(lst,x))
```

## W6A24 
Viêết hàm đẩy tất cả các số 0 về đầu (về cuối), các số khác vẫn đúng vị trí 
trước sau: 0 2 3 2 4 0 0 0 9 => 0 0 0 2 3 2 4 9 

```python
def zero_to_top(lst):
    for i in range(len(lst)):
        if lst[i] == 0:
            lst.insert(0, lst[i])
            lst.pop(i+1)
    return lst
lst = list(map(int, input().split()))
print(*zero_to_top(lst))
```

## W6A26 
Tính tổng các phần tử cực đại địa phương trong một danh sách, không tính 2 
phần tử đầu và cuối 

```python
lst = list(map(int, input().split()))
tong = 0
for i in range(1, len(lst)-1):
    if lst[i-1]<lst[i] and lst[i+1]<lst[i]:
        tong += lst[i]
print(tong)
```

## W6A27 
Cho một dãy số nguyên, Tìm và in ra vị trí các số < x (x là giá trị cho trước). 
nếu không có in ra -1 

```python
lst = list(map(int, input().split()))
x = int(input())
kq = []
for i in range(len(lst)):
    if lst[i]<x:
        kq.append(i)
if len(kq) == 0:
    print(-1)
else:
    print(*kq)
```
