## W4A18 Nhập vào số nguyên dương a và b, in toàn bộ ước chung của a và b 

```python
a, b = map(int, input().split())
for i in range(1,min(a,b)+1):
    if a%i==0 and b%i == 0:
        print(i, end=" ")
```

Viết chương trình Python sử dụng một trong 2 loại vòng lặp for hoặc while để in ra các số 
dương chẵn nhỏ hơn hoặc bằng n (n <= 1000) 

```python
n = int(input())
if n>0 and n<=1000:
    for i in range(2,n+1):
        if i%2==0:
            print(i, end=" ")
else:
    print("ValueError")
```

## W4A24 
Nhập vào A, tìm n nhỏ nhất sao cho 
1 + 1/2 + 1/3 + 1/4 + ... + 1/n > A 

```python
A = int(input())
s = 0
n = 1
while s+1/n<A:
    s += 1/n
    n += 1
print(n-1)
```

## W4A25 
Nhập vào một dãy số nguyên, ngưng nhập khi người dùng nhập -1. 
Sau khi nhập xong, in số lớn nhất, số nhỏ nhất trong những số vừa nhập

```python
n = int(input())
danh_sach = []
while n != -1:
    danh_sach.append(n)
    n = int(input())
print(f" max la {max(danh_sach)}")
print(f" min là {min(danh_sach)}")
    
```

## W4A26 
[Fibonaci1] 
Dãy số fibonacci là dãy số được định nghĩa như sau: 1, 1, 2, 3, 5, 8, 13,... với số kế tiếp sẽ 
bằng tổng hai số trước đó 
Nhập vào số nguyên dương A, hãy tìm số trong dãy số fibonacci lớn nhất nhưng không vượt 
quá A 

```python
A = int(input())
f1= 1
f2 = 1
f_tong = 0
while f1+f2<=A:
    f_tong = f1 + f2
    f1 = f2
    f2 = f_tong
print(f2)
```

## W4A28 
Nhập vào một chuỗi, hãy in từ đầu tiên trong chuỗi 
Dùng cách: for trên index; KHÔNG dùng hàm split() 

```python
chuoi = input()
tu_dau_tien = ""
for i in range(len(chuoi)):
    ki_tu_dau_tien = chuoi[i]
    if ki_tu_dau_tien != " ":
        tu_dau_tien += ki_tu_dau_tien
    elif len(tu_dau_tien)>0:
        break
print(tu_dau_tien)
```

## W4A29

## W4A31 
Nhập vào một chuỗi, hãy đếm xem trong chuỗi có bao nhiêu ký tự in hoa, bao nhiêu ký tự in 
thường, bao nhiêu ký tự số

```python
n = input()
so_in_hoa = 0
so_in_thuong = 0
so_so = 0
ki_tu = ""
for i in range(len(n)):
    ki_tu = n[i]
    if ki_tu.isupper():
        so_in_hoa += 1
    elif ki_tu.islower():
        so_in_thuong += 1
    elif ki_tu.isdigit():
        so_so += 1
print(f"{so_in_hoa} {so_in_thuong} {so_so}")
```
